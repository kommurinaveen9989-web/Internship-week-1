```python
# Defining the dataset based on the assignment sheet
# Structure: [Name, Subject A, Subject B, Subject C, Subject D, Subject E]
dataset = [
    ["Ankit", 41,34,45,55,63],
    ["Aravind",42,23,34,44,53],
    ["Lakshay",32,23,13,54,67],
    ["Gyan",23,82,23,63,34],
    ["Pranav",21,23,25,56,56]
]
```

```python
# Variables to track the highest score and the top scorer's name
max_marks_b = -1
top_scorer_b = ""

for row in dataset:
    name = row[0]
    marks_b = row[2]  # Subject B is at index 2

    if marks_b > max_marks_b:
        max_marks_b = marks_b
        top_scorer_b = name

print(f"Highest marks in Subject B: {max_marks_b} scored by {top_scorer_b}")
```

```python
# Calculating total marks for Subject C to find the average
total_marks_c = 0
num_candidates = len(dataset)

for row in dataset:
    total_marks_c += row[3]  # Subject C is at index 3

average_c = total_marks_c / num_candidates
print(f"Average marks scored in Subject C: {average_c:.2f}")
```

```python
# Variables to track the highest overall percentage
highest_overall_pct = -1.0
top_overall_candidate = ""

for row in dataset:
    name = row[0]
    # Summing marks from index 1 to 5
    total_marks = sum(row[1:6])
    percentage = (total_marks / 500) * 100

    if percentage > highest_overall_pct:
        highest_overall_pct = percentage
        top_overall_candidate = name

print(f"Highest overall percentage: {highest_overall_pct:.2f}% by {top_overall_candidate}")
```

```python
# Variables to track the highest top-4 percentage
highest_top4_pct = -1.0
top_four_candidate = ""

for row in dataset:
    name = row[0]
    marks_list = row[1:6]

    # Exclude the minimum mark to get the top 4 scores
    top_4_marks = sorted(marks_list, reverse=True)[:4]
    sum_top4 = sum(top_4_marks)

    top4_percentage = (sum_top4 / 400) * 100

    if top4_percentage > highest_top4_pct:
        highest_top4_pct = top4_percentage
        top_four_candidate = name

print(f"Highest top-4 percentage: {highest_top4_pct:.2f}% by {top_four_candidate}")
```
