```python
# Kommuri_Naveen_Python_Beginner.ipynb
# PROBLEM 1: Fibonacci Series (First N Numbers)
#  Iterative Approach
def fibonacci_iterative(n):
    """Prints the first N Fibonacci numbers iteratively."""
    if n <= 0:
        return []
    elif n == 1:
        return [0]

    series = [0, 1]
    while len(series) < n:
        series.append(series[-1] + series[-2])
    return series

# Recursive Approach Helper
def fibonacci_recursive_helper(n):
    if n <= 0:
        return 0
    if n == 1:
        return 1
    return fibonacci_recursive_helper(n - 1) + fibonacci_recursive_helper(n - 2)

def print_fibonacci_recursive(n):
    """Returns the first N Fibonacci numbers using recursion."""
    return [fibonacci_recursive_helper(i) for i in range(n)]

# Testing the functions
num_terms = 7
print(f"Iterative Fibonacci ({num_terms} terms): {fibonacci_iterative(num_terms)}")
print(f"Recursive Fibonacci ({num_terms} terms): {print_fibonacci_recursive(num_terms)}")
```

```python
# PROBLEM 2: Palindrome Checker (String or Number)

def is_palindrome(value):
    """Checks if a string or number reads the same backward as forward."""
    # Convert input to a lowercase string and remove any spaces
    clean_str = str(value).lower().replace(" ", "")

    # Compare the string with its reverse using slicing
    return clean_str == clean_str[::-1]

# Testing Problem 2
print(f"Is 'radar' a palindrome?: {is_palindrome('radar')}")
print(f"Is 12321 a palindrome?: {is_palindrome(12321)}")
print(f"Is 'hello' a palindrome?: {is_palindrome('hello')}")
```

```python
# PROBLEM 3: Factorial of a Number
# Iterative Approach
def factorial_iterative(n):
    """Computes factorial using a standard loop."""
    result = 1
    # Loop from 1 up to n to multiply the values
    for i in range(1, n + 1):
        result *= i
    return result

# Recursive Approach
def factorial_recursive(n):
    """Computes factorial using recursion."""
    # Base case: factorial of 0 or 1 is 1
    if n == 0 or n == 1:
        return 1
    # Recursive case: n * factorial of (n-1)
    return n * factorial_recursive(n - 1)

# Testing Problem 3
num = 5
print(f"Factorial of {num} (Iterative): {factorial_iterative(num)}")
print(f"Factorial of {num} (Recursive): {factorial_recursive(num)}")
```

```python
# PROBLEM 4: Prime Number Check & Primes 1-100


def is_prime(n):
    """Helper function to determine if a single number is prime."""
    if n <= 1:
        return False
    # Check for factors up to the square root of n for efficiency
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False  # Factor found, not prime
    return True

# Generate and print all prime numbers between 1 and 100
primes_1_to_100 = [num for num in range(1, 101) if is_prime(num)]

# Testing Problem 4
print(f"Is 29 prime?: {is_prime(29)}")
print(f"Primes between 1 and 100:\n{primes_1_to_100}")
```

```python
# PROBLEM 5: Armstrong Number Checker

def is_armstrong(n):
    """Checks if a number equals the sum of its digits raised to the power of the total digit count."""
    num_str = str(n)
    num_digits = len(num_str)

    # Sum each digit raised to the power of the total number of digits
    total_sum = sum(int(digit) ** num_digits for digit in num_str)

    return total_sum == n

# Testing Problem 5
print(f"Is 153 an Armstrong number?: {is_armstrong(153)}")
print(f"Is 9474 an Armstrong number?: {is_armstrong(9474)}")
print(f"Is 123 an Armstrong number?: {is_armstrong(123)}")
```

```python
# PROBLEM 6: Reverse a Number & Sum of Digits

def reverse_and_sum(n):
    """Reverses an integer and calculates the sum of its digits."""
    # Work with absolute value to handle string conversion easily
    num_str = str(abs(n))

    # Reverse using string slicing
    reversed_num = int(num_str[::-1])
    if n < 0:
        reversed_num = -reversed_num

    # Calculate digit sum by iterating through characters
    digit_sum = sum(int(digit) for digit in num_str)

    return reversed_num, digit_sum

# Testing Problem 6
number = 4562
rev, d_sum = reverse_and_sum(number)
print(f"Original: {number} | Reversed: {rev} | Sum of Digits: {d_sum}")
```

```python
# PROBLEM 7: Find the Second Largest Number


def find_second_largest(numbers):
    """Finds the second distinct largest number in a list without built-in sorting."""
    if len(numbers) < 2:
        return None

    first = second = float('-inf')
    for num in numbers:
        # If number is larger than first, update both first and second
        if num > first:
            second = first
            first = num
        # If it's between first and second, update second
        elif num > second and num != first:
            second = num

    return second if second != float('-inf') else None

# Testing Problem 7
my_list = [12, 35, 1, 10, 34, 35, 1]
print(f"Second largest in {my_list} is: {find_second_largest(my_list)}")
```

```python
# PROBLEM 8: Count Vowels and Consonants

def count_vowels_consonants(s):
    """Counts vowels and consonants in a string, ignoring spaces/symbols."""
    vowels = "aeiou"
    v_count = 0
    c_count = 0

    for char in s.lower():
        if char.isalpha():  # Check if character is a valid alphabet letter
            if char in vowels:
                v_count += 1
            else:
                c_count += 1

    return {"Vowels": v_count, "Consonants": c_count}

# Testing Problem 8
text = "Vel Tech University"
print(f"Counts for '{text}': {count_vowels_consonants(text)}")
```

```python
# PROBLEM 9: Find the Missing Number (1 to N)


def find_missing_number(arr, n):
    """Finds the missing number using the natural sum formula: N*(N+1)//2."""
    expected_sum = (n * (n + 1)) // 2
    actual_sum = sum(arr)

    # The difference gives the single missing number
    return expected_sum - actual_sum

# Testing Problem 9 (Range 1 to 6, missing 4)
test_arr = [1, 2, 3, 5, 6]
print(f"The missing number is: {find_missing_number(test_arr, 6)}")
```

```python
# PROBLEM 10: Anagram Checker


def is_anagram(str1, str2):
    """Checks if two strings are anagrams of each other."""
    # Strip spaces and make lowercase
    clean1 = str1.lower().replace(" ", "")
    clean2 = str2.lower().replace(" ", "")

    if len(clean1) != len(clean2):
        return False

    # Anagrams will have identical sorted character sequences
    return sorted(clean1) == sorted(clean2)

# Testing Problem 10
s1, s2 = "Listen", "Silent"
print(f"Are '{s1}' and '{s2}' anagrams?: {is_anagram(s1, s2)}")
```
